# Step 1 - Apply standard scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
df = sc.fit_transform(df)
```

# Step 2 - Find Covariance Matrix

# Step 3 - Finding EV and EVs

```python
covariance_matrix = np.cov(df, rowvar=False)
eigen_values, eigen_vectors = np.linalg.eig(covariance_matrix)
print('Covariance Matrix:\n shape', len(covariance_matrix))
```

```python
pc = eigen_vectors[0:5]
len(pc[0])
```

```python
transformed_df = np.dot(df,pc.T)
new_df = pd.DataFrame(transformed_df,columns=['PC1','PC2','PC3','PC4','PC5'])
#new_df['class_name'] = dff['class_name'].values
new_df.head()
```

```python
new_df.shape
```

```python
rec = np.dot(new_df,pc)
```

```python
rec.shape
```

```python
rec = pd.DataFrame(rec)
```

```python
rec.sample(5)
```

```python
reconstruction_error = np.mean((df- rec)**2)
print(reconstruction_error)
```
