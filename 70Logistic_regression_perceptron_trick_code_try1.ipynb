```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/logistic_csv.csv')
x = df['X']
y = df['Y']
df.shape
```

```python
sns.scatterplot(data=df,x=x,y=y,hue=df['Output'])
```

#logistic regression from scratch

```python
# ax + by + c = 0
a = 1
b = 50
c = 1
lr = 0.01 # learning rate
sns.scatterplot(data=df,x=x,y=y,hue=df['Output']) # ploting data points
liney = -(a*x + c) / b # y value
sns.lineplot(x=x, y=liney,color='red') # initial line -> which is in red color
for ep in range(500):
    for i in range(len(df)):
        current_point_x = df.iloc[i, 0]
        current_point_y = df.iloc[i, 1]
        if a*current_point_x + b * current_point_y + c < 0:
            yhat = 0
        else:
            yhat = 1
        error = df.iloc[i]['Output']-yhat
        #print(error)
        a = a+(lr*current_point_x*(error))
        b = b+(lr*current_point_y*(error))
        c = c+(lr*(error))

liney = -(a*x + c) / b
sns.lineplot(x=x, y=liney,color='green')
plt.show()
```
