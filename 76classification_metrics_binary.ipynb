#logistic metrics for two classes

```python
import numpy as np
import pandas as pd

np.random.seed(42)
n_rows = 500
n_features = 10

df = np.random.normal(0, 1, (n_rows, n_features))
weights = np.array([1.2, -1.0, 0.8, 0.6, -0.7, 0.5, 0.4, -0.3, 0.9, -0.5])
logits = df @ weights + np.random.normal(0, 0.8, n_rows)
probabilities = 1 / (1 + np.exp(-logits))
y = (np.random.random(n_rows) < probabilities).astype(int)

df = pd.DataFrame(df, columns=[f"Feature{i}" for i in range(1, 11)])
```

```python
df.head()
```

#train test split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

#training logistic and decision tree

```python
clf1 = LogisticRegression()
clf2 = DecisionTreeClassifier()
```

```python
clf1.fit(X_train,y_train)
clf2.fit(X_train,y_train)
```

#predicting

```python
y_pred1 = clf1.predict(X_test)
y_pred2 = clf2.predict(X_test)
```

#calculating accuracy score

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print("Accuracy of Logistic Regression",accuracy_score(y_test,y_pred1))
print("Accuracy of Decision Trees",accuracy_score(y_test,y_pred2))
```

#calculating confusion matrix

```python
confusion_matrix(y_test,y_pred1)
```

```python
print("Logistic Regression Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred1),columns=list(range(0,2)))
```

```python
print("Decision Tree Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred2),columns=list(range(0,2)))
```

```python
result = pd.DataFrame()
result['Actual Label'] = y_test
result['Logistic Regression Prediction'] = y_pred1
result['Decision Tree Prediction'] = y_pred2
```

```python
result.sample(10)
```

#calculating recall, precision, f1 score

```python
from sklearn.metrics import recall_score,precision_score,f1_score
```

```python
print("For Logistic regression Model")
print("-"*50)
cdf = pd.DataFrame(confusion_matrix(y_test,y_pred1),columns=list(range(0,2)))
print(cdf)
print("-"*50)
print("Precision - ",precision_score(y_test,y_pred1))
print("Recall - ",recall_score(y_test,y_pred1))
print("F1 score - ",f1_score(y_test,y_pred1))
```

```python
print("For DT Model")
print("-"*50)
cdf = pd.DataFrame(confusion_matrix(y_test,y_pred2),columns=list(range(0,2)))
print(cdf)
print("-"*50)
print("Precision - ",precision_score(y_test,y_pred2))
print("Recall - ",recall_score(y_test,y_pred2))
print("F1 score - ",f1_score(y_test,y_pred2))
```
