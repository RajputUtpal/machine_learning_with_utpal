### `Task` Train a KNN model on glass type dataset and find best n_neighnours.

Data Link: https://drive.google.com/file/d/17cbDNBmys04MJqQfrma3jd72VPMnxIq0/view?usp=share_link

```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
```

```python
df = pd.read_csv('/content/glass.csv')
```

```python
X = df.iloc[:,0:-1]
y = df.iloc[:,-1]
```

```python
sc = StandardScaler()
X = sc.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=20,test_size=0.2)
```

```python
def knnf(i):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train,y_train)
    ypred = knn.predict(X_test)
    return accuracy_score(y_test,ypred)
```

```python
accuracy = []
for i in range (1,16):
    i = int(i)
    accuracy.append(knnf(i))
import matplotlib.pyplot as plt
plt.plot(range(1,16),accuracy)
```
