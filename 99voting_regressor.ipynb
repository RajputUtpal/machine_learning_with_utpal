#imports

```
from sklearn.datasets import fetch_california_housing as load_boston
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

#load dataset

```
X,y = load_boston(return_X_y=True)
```

```
X.shape,y.shape
```

#making objects of 3 different models

```
lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
```

#estimator

```
estimators = [('lr',lr),('dt',dt),('svr',svr)]
```

#voting regressor

```
from sklearn.ensemble import VotingRegressor
```

```
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=3)
print("Voting Regressor",np.round(np.mean(scores),2))
```

#voting regressor with weights

```
for i in range(1,3):
  for j in range(1,3):
    for k in range(1,3):
      vr = VotingRegressor(estimators,weights=[i,j,k])
      scores = cross_val_score(vr,X,y,scoring='r2',cv=3)
      print("For i={},j={},k={}".format(i,j,k),np.round(np.mean(scores),2))
```

#5 objects/model of decison tree with different depths

```
# using the same algorithm

dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
```

#estimators

```
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
```

#calculating scores of all 5 models individually, so that we would be able to compare the net performance increase

```
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=3)
  print(estimator[0],np.round(np.mean(scores),2))
```

#cross validation of voting regression

```
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=3)
print("Voting Regressor",np.round(np.mean(scores),2))
```
