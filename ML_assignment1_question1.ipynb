```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Given data
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([45, 50, 55, 65, 70], dtype=float)
```

```python
plt.scatter(X,y)
```

```python
# Initialize parameters
m = 0.0
b = 0.0

# Hyperparameters
learning_rate = 0.01
epochs = 1000
```

```python

# Gradient Descent
n = len(X)

for i in range(epochs):
    # Predictions
    y_pred = m * X + b

    # Calculate gradients
    dm = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Update parameters
    m = m - learning_rate * dm
    b = b - learning_rate * db

# Predict marks for 6 hours
hours = 6
prediction = m * hours + b

print("Slope (m):", m)
print("Intercept (b):", b)
print("Regression equation: y =", m, "* x +", b)
print("Predicted marks for 6 hours:", prediction)
```

#Using Class

```python

import numpy as np

class SimpleLinearRegression:
    def __init__(self, lr=0.01, epochs=1000):
        self.lr = lr
        self.epochs = epochs
        self.b0 = 0.0
        self.b1 = 0.0

    def fit(self, X, y):
        X = np.asarray(X, dtype=float)
        y = np.asarray(y, dtype=float)
        n = len(y)

        for _ in range(self.epochs):
            y_pred = self.b0 + self.b1 * X
            error = y_pred - y

            db0 = (2/n) * np.sum(error)
            db1 = (2/n) * np.sum(error * X)

            self.b0 -= self.lr * db0
            self.b1 -= self.lr * db1

    def predict(self, X):
        X = np.asarray(X, dtype=float)
        return self.b0 + self.b1 * X
```

```python
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([45, 50, 55, 65, 70], dtype=float)

model = SimpleLinearRegression(lr=0.01, epochs=5000)
model.fit(X, y)

print("Intercept:", model.b0)
print("Slope:", model.b1)
print("Prediction for x=6:", model.predict([6]))
```

```python
plt.scatter(X,y)
plt.plot(X,model.predict(X),color='red')
plt.grid()
plt.xlabel('Hour')
plt.ylabel('Marks')
```
