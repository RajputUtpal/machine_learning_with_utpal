```python
import pandas as pd
import numpy as np
df = pd.read_csv('/content/1309288767715_spam_nonspam_50_samples.csv')
df.sample(5)
```

```python
df1 = df[df['Class'] == 'Spam']
df2 = df[df['Class'] == 'Not Spam']
```

```python
prior_spam = df1.shape[0]/df.shape[0]
prior_spam
```

```python
prior_not_spam = df2.shape[0]/df.shape[0]
prior_not_spam
```

```python
x = ["Report",	"Work",	"Many"]
```

#without smoothing

```python
#spam
p0 = df1[df1['Subject']== x[0]].shape[0]/df1.shape[0]
p1 = df1[df1['SenderType']== x[1]].shape[0]/df1.shape[0]
p2 = df1[df1['Links']== x[2]].shape[0]/df1.shape[0]
prob_spam_x = p0*p1*p2*prior_spam
prob_spam_x
```

```python
#not spam
p0 = df2[df2['Subject']== x[0]].shape[0]/df2.shape[0]
p1 = df2[df2['SenderType']== x[1]].shape[0]/df2.shape[0]
p2 = df2[df2['Links']== x[2]].shape[0]/df2.shape[0]
prob_not_spam_x = p0*p1*p2*prior_not_spam
prob_not_spam_x
```

```python
if prob_spam_x > prob_not_spam_x:
    print("Spam")
elif prob_spam_x = prob_not_spam_x:
    print("same probability")
else:
    print("Not Spam")
```

#probability using laplaced smoothing

```python
#spam
p0 = (df1[df1['Subject']== x[0]].shape[0]+1)/(df1.shape[0]+df1['Subject'].unique().shape[0])
p1 = (df1[df1['SenderType']== x[1]].shape[0]+1)/(df1.shape[0]+df1['SenderType'].unique().shape[0])
p2 = (df1[df1['Links']== x[2]].shape[0]+1)/(df1.shape[0]+df1['Links'].unique().shape[0])
prob_spam_x = p0*p1*p2*prior_spam
prob_spam_x
```

```python
#not spam
p0 = (df2[df2['Subject']== x[0]].shape[0]+1)/(df2.shape[0]+df2['Subject'].unique().shape[0])
p1 = (df2[df2['SenderType']== x[1]].shape[0]+1)/(df2.shape[0]+df2['SenderType'].unique().shape[0])
p2 = (df2[df2['Links']== x[2]].shape[0]+1)/(df2.shape[0]+df2['Links'].unique().shape[0])
prob_spam_x = p0*p1*p2*prior_spam
prob_spam_x
```

```python
if prob_spam_x > prob_not_spam_x:
    print("Spam")
elif prob_spam_x = prob_not_spam_x:
    print("same probability")
else:
    print("Not Spam")
```
