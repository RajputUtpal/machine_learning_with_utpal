#imports and read csv

```python
import pandas as pd
import numpy as np
from scipy.stats import norm
df = pd.read_csv('/content/1310425495534_tennis_numerical_50_samples.csv')
df.sample(5)
```

#divide in df1,df2

```python
df1 = df[df['Play'] == 'Yes']
df2 = df[df['Play'] == 'No']
```

#mean

```python
df1_mean = df1.iloc[:,0:3].mean()
df2_mean = df2.iloc[:,0:3].mean()
```

#std

```python
df1_std = df1.iloc[:,0:3].std()
df2_std = df2.iloc[:,0:3].std()
```

#prior yes

```python
prior_yes = df1.shape[0]/df.shape[0]
prior_yes
```

#prior no

```python
prior_no = df2.shape[0]/df.shape[0]
prior_no
```

#choosing a x from that data set only

```python
x = df.sample()
ans = x.values.ravel()[-1]
x = x.values[:,0:3].ravel()
x,ans
```

#probability of yes when x is given

```python
#yes
p0 = norm.pdf(x[0], df1_mean.iloc[0], df1_std.iloc[0]) # p0 = probability of x[0] in class "yes"
p1 = norm.pdf(x[1], df1_mean.iloc[1], df1_std.iloc[1])
p2 = norm.pdf(x[2], df1_mean.iloc[2], df1_std.iloc[2])
prob_yes_x = p0*p1*p2*prior_yes
prob_yes_x
```

#probability of no when x is given

```python
#no2
p0 = norm.pdf(x[0], df2_mean.iloc[0], df2_std.iloc[0])
p1 = norm.pdf(x[1], df2_mean.iloc[1], df2_std.iloc[1])
p2 = norm.pdf(x[2], df2_mean.iloc[2], df2_std.iloc[2])
prob_no_x = p0*p1*p2*prior_no
prob_no_x
```

comparing posterior

```python
print(x)
if prob_yes_x > prob_no_x:
    print('predicted= Yes\u2705')
else:
    print('predicted= No\u274C')
print('actual=',ans)
```

#checking accuracy

```python
count = 0
for i in range(20):
    x = df.sample()
    ans = x.values.ravel()[-1]
    x = x.values[:,0:3].ravel()
    p0 = norm.pdf(x[0], df1_mean.iloc[0], df1_std.iloc[0])
    p1 = norm.pdf(x[1], df1_mean.iloc[1], df1_std.iloc[1])
    p2 = norm.pdf(x[2], df1_mean.iloc[2], df1_std.iloc[2])
    prob_yes_x = p0*p1*p2*prior_yes
    prob_yes_x
    p0 = norm.pdf(x[0], df2_mean.iloc[0], df2_std.iloc[0])
    p1 = norm.pdf(x[1], df2_mean.iloc[1], df2_std.iloc[1])
    p2 = norm.pdf(x[2], df2_mean.iloc[2], df2_std.iloc[2])
    prob_no_x = p0*p1*p2*prior_no
    prob_no_x
    print("X=",x)
    if prob_yes_x > prob_no_x:
        pred = 'Yes'
        print('predicted= Yes')
    else:
        pred = 'No'
        print('predicted= No')
    print('actual=',ans)
    if pred == ans:
        print('prediction is \u2705')
    else:
        print('prediction is \u274C')
    print('\n')
```
